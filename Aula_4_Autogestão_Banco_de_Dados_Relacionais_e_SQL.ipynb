## Parte 2 - Prática em SQLite com Python

Usando Python e o pacote sqlite3:

a. Crie o banco de dados local `cinema.db` e adicione as tabelas filmes e avaliacoes com os campos definidos acima.

b. Implemente as quatro operações CRUD para ambas tabelas, ou seja, insira, leia, atualize e remova registros diretamente pelo Python.

```python
import sqlite3
```

```python
con = sqlite3.connect("cinema.db")
cur = con.cursor()
```

## 1. Craindo tabelas

```python
cur.execute("""CREATE TABLE filmes (
  id INTEGER PRIMARY KEY AUTOINCREMENT,
  titulo TEXT NOT NULL,
  genero TEXT NOT NULL,
  ano_lancamento INTEGER
);""")

cur.execute("""CREATE TABLE avaliacoes(
	id INTEGER PRIMARY KEY AUTOINCREMENT,
  	filme_id INTEGER NOT NULL REFERENCES filmes(id) ON DELETE CASCADE,
  	nota INTEGER CHECK(nota>=1 and nota<=5),
  	comentario TEXT
);""")
```

```python
con.commit()
```

## 2. CRUD

```python
#Inserindo dados nas tabelas

filmes =[
('Inception', 'Ação', 2010),
('Toy Story 3', 'Animação', 2010),
('Titanic', 'Drama', 1997),
('Gladiador', 'Ação', 2000),
('O Rei Leão', 'Animação', 1994),
('Parasita', 'Drama', 2019),
('Matrix', 'Ficção', 1999),
('Vingadores: Ultimato', 'Ação', 2019),
('Forrest Gump', 'Drama', 1994),
('Coringa', 'Drama', 2019),
('Divertida Mente', 'Animação', 2015),
('Shrek', 'Animação', 2001),
('Se Beber, Não Case!', 'Comédia', 2009),
('Superbad', 'Comédia', 2007),
('O Poderoso Chefão', 'Drama', 1972),
('Avatar', 'Ficção', 2009),
('Interestelar', 'Ficção', 2014),
('Homem-Aranha: Sem Volta para Casa', 'Ação', 2021),
('Frozen', 'Animação', 2013),
]

cur.executemany("INSERT INTO filmes (titulo, genero, ano_lancamento) VALUES (?, ?, ?)", filmes)
cur.execute("INSERT INTO filmes (titulo, genero, ano_lancamento) VALUES (?, ?, ?)", ('Up: Altas Aventuras', 'Animação', 2009))

con.commit()
```

```python
#Consultando filmes
def listar_filmes():
  print("Filmes:")
  for row in cur.execute("SELECT * FROM filmes"):
    print(row)
```

```python
#Atualizar filme
def atualizar_ano(id, novo_ano):
  cur.execute("UPDATE filmes SET ano_lancamento = ? WHERE id = ?", (novo_ano, id))
  con.commit()
```
