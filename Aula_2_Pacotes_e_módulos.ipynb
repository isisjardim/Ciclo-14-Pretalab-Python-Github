# Módulos e Pacotes em Python

## Por que módulos e pacotes?
Reaproveitar código acelera o desenvolvimento e evita "reinventar a roda".

Um **módulo** é um arquivo `.py` com definições (funções, classes, variáveis).

Um **pacote** é uma pasta com módulos (geralmente contém um `__init__.py`).

**Importar** significa "trazer" nomes de outro módulo para usar aqui.

## Biblioteca padrão
Python vem com muitos módulos úteis já como padrão, alguns exemplos são:
- `math`: funções matemáticas
- `random`: aleatoriedade
- `os` / `pathlib`: arquivos e diretórios
- `datetime`: datas e horários

## Sintaxe de import
Para usar um módulo utilizamos a palavra `import`:

- `import modulo`
- `import modulo as apelido`
- `from modulo import nome1, nome2`
- `from modulo import *` (pode causar conflitos e não é uma boa prática)

```python
import math
import math as m
from math import sqrt, pi, log

print("math.pi:", math.pi)
print("m.e:", m.e)
print("sqrt(16):", sqrt(16))
print(log(60))
print("pi:", pi)
```

```python
import random

print("Número aleatório:", random.randint(2,20))

#Adicionando numero a uma lista
lista_numeros = []
lista_numeros.append(random.randint(1,99))
lista_numeros.append(random.randint(1,99))
```

### Funções úteis:
- `max(lista)` → retorna o maior valor.

- `min(lista)` → retorna o menor valor.

- `sum(lista)` → soma os valores (ótimo se quiser calcular média).

- `len(lista)` → quantidade de elementos.

```python
import random

lista_aleatoria = []
lista_aleatoria.append(random.randint(1,100))
lista_aleatoria.append(random.randint(1,100))
lista_aleatoria.append(random.randint(1,100))
lista_aleatoria.append(random.randint(1,100))
lista_aleatoria.append(random.randint(1,100))

print("lista:", lista_aleatoria)
print("Número máximo:", max(lista_aleatoria))
print("Número mínimo:", min(lista_aleatoria))
print("Soma:", sum(lista_aleatoria))
print("Tamanho da lista:", len(lista_aleatoria))
```

### Exercício 2 — Seu módulo
1. Crie um módulo `calculadora.py` com as funções:
- area_retangulo(l, a) → retorna a área de um retângulo de lados l e a.

- perimetro_retangulo(l, a) → retorna o perímetro de um retângulo de lados l e a.

- area_circulo(r) → retorna a área de um círculo de raio r.

- circunferencia(r) → retorna o comprimento da circunferência de raio r.

- volume_cubo(l) → retorna o volume de um cubo de lado l.

2. Importe o módulo e **teste** as funções.

👉 Dica: use math.pi quando precisar de π.

```python
%%writefile calculadora.py
from math import pi

def area_retangulo(l, a):
  return l*a

def area_quadrado(l):
  return l*l

def perimetro_retangulo(l,a):
  return 2*(l+a)

def area_circulo(r):
  return pi*(r**2)

def circunferencia(r):
  return 2*pi*r

def volume_cubo(l):
  return l**3
```

```python
import calculadora as cl

print(cl.area_retangulo(2,3))
print(cl.perimetro_retangulo(2,3))

print(cl.area_quadrado(4))

print(cl.area_circulo(8))
print(cl.circunferencia(8))

print(cl.volume_cubo(9))
```

### Boas práticas
- Nomeie módulos com **minúsculas** (ex.: `utilidades.py`).
- Evite `from X import *` — seja explícito.
- Agrupe funções relacionadas;
- Use `requirements.txt` (ou seção de dependências) em projetos reais.
- Ao editar um módulo no Colab, use `importlib.reload(modulo)` após `%%writefile`.

## Pacotes `requests` e `BeautifulSoup`

Para extração de dados.

```python
import requests
from bs4 import BeautifulSoup

def extrair_noticia(url):
  # Fazendo a requisição
  resposta = requests.get(url)

  if resposta.status_code == 200:
      # Parse do HTML
      soup = BeautifulSoup(resposta.text, "html.parser")

      # Pegando o título (geralmente está em <h1>)
      titulo = soup.find("h1")

      print("Título da notícia:")
      print(titulo.get_text() if titulo else "Não encontrado")

      # Pegando os primeiros parágrafos (geralmente em <p>)
      paragrafos = soup.find_all("p")
      print("\nPrimeiros parágrafos:")
      for p in paragrafos[:3]:
          print("-", p.get_text())
  else:
      print("Erro ao acessar a página:", resposta.status_code)
```

```python
url_bbc = "https://www.bbc.com/portuguese/articles/cp98231le9mo"
extrair_noticia(url_bbc)
```

```python
url_g1 = "https://g1.globo.com/ciencia/noticia/2025/09/16/mumias-fumaca-cientistas-apontam-as-mumificacoes-mais-antigas-do-mundo-asia.ghtml"
extrair_noticia(url_g1)
```

```python
url_govbr= "https://www.gov.br/pt-br/servicos/consultar-estabelecimentos-que-possuem-o-selo-turismo-responsavel"
extrair_noticia(url_govbr)
```
